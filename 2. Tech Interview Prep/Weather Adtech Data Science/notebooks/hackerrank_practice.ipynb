```python
# Synthetic churn dataset to stand in for a real "provided dataset"
rng = np.random.default_rng(3)
n = 800

df_model = pd.DataFrame({
    "tenure_months": rng.integers(1, 72, n),
    "monthly_spend": np.round(rng.normal(70, 25, n), 2).clip(min=5),
    "num_support_tickets": rng.poisson(1.5, n),
    "contract_type": rng.choice(["Monthly", "Annual", "Two-Year"], n, p=[0.5, 0.3, 0.2]),
    "region": rng.choice(["East", "West", "North", "South"], n),
})

# construct a churn label with a real (learnable) signal + noise
logit = (
    -1.5
    + 0.03 * (36 - df_model["tenure_months"])
    + 0.01 * df_model["monthly_spend"]
    + 0.35 * df_model["num_support_tickets"]
    + (df_model["contract_type"] == "Monthly") * 0.8
)
prob = 1 / (1 + np.exp(-logit))
df_model["churned"] = (rng.random(n) < prob).astype(int)

df_model.head()
```
